```python
import os
from dotenv import load_dotenv
from unsloth import FastVisionModel

load_dotenv()
hf_token = os.getenv("HF_TOKEN")

if not hf_token:
    raise ValueError("❌ Error: HF_TOKEN not found in .env or environment variables!")

model, tokenizer = FastVisionModel.from_pretrained(
    "bbarni/qwen-3vl-experiment",
    load_in_4bit=False,
)

# This does the merge, save, and push automatically
model.push_to_hub_merged(
    "bbarni/qwen-3vl-merged",
    tokenizer,
    save_method="merged_16bit", # Forces standardized float16 (Best for vLLM)
    token=hf_token
)

print("✅ Upload Complete!")
```
