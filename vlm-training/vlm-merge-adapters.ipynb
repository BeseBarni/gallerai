```python
from unsloth import FastVisionModel

# 1. Load your specific trained model
# Replace with your local path or HuggingFace ID
model, tokenizer = FastVisionModel.from_pretrained(
    "bbarni/qwen-3vl-experiment", 
    load_in_4bit=False, # IMPORTANT: Must load in 16bit to merge!
)

# 2. Merge the LoRA adapters into the base model
print("Merging model...")
model.merge_and_unload()

# 3. Save the "baked" model to a new folder
output_dir = "merged_qwen_vl"
model.save_pretrained(output_dir)
tokenizer.save_pretrained(output_dir)

print(f"Done! Model saved to {output_dir}")
```
